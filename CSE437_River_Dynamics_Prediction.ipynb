```python
import numpy as np
import pandas as pd

def make_unique(series):
    """
    Make all values in a Series unique by adding small increments to duplicates.
    """
    seen = {}
    result = []

    for val in series:
        if pd.isnull(val):

            result.append(val)
            continue

        if val not in seen:
            seen[val] = 0
            result.append(val)
        else:
            seen[val] += 1

            new_val = val + seen[val] * 0.0001
            result.append(new_val)
    return pd.Series(result, index=series.index)

columns_to_fix = ['mean_ndwi', 'mean_savi', 'percent_builtup', 'percent_vegetation']

for col in columns_to_fix:
    df1[col] = make_unique(df1[col])

for col in columns_to_fix:
    num_duplicates = df1[col].duplicated().sum()
    print(f"Duplicates in '{col}': {num_duplicates}")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    AdaBoostClassifier
)
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from imblearn.over_sampling import ADASYN
from xgboost import XGBClassifier

# Assume df1 is already loaded and cleaned
X = df1.drop(columns=[
    "erosion_rate_class_encoded",
    "river_name_encoded",
    "erosion_rate_label",
    "river_name_label",
    "year"

])
y = df1["erosion_rate_class_encoded"]

# Feature scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-test split with stratify
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

# Apply ADASYN to training data
adasyn = ADASYN(random_state=42)
X_train_res, y_train_res = adasyn.fit_resample(X_train, y_train)

# Define classifiers including AdaBoost and XGBoost
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "AdaBoost": AdaBoostClassifier(random_state=42),
    "SVM": SVC(),
    "Naive Bayes": GaussianNB(),
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "XGBoost": XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=42)
}

results = {}
reports = {}

# Train and evaluate each model on ADASYN-resampled data
for name, model in models.items():
    model.fit(X_train_res, y_train_res)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    results[name] = acc
    reports[name] = classification_report(y_test, y_pred, output_dict=True)

# Convert accuracy results to DataFrame
results_df = pd.DataFrame.from_dict(results, orient="index", columns=["Accuracy"]).sort_values(by="Accuracy", ascending=False)

# Plot model comparison
plt.figure(figsize=(10, 6))
sns.barplot(x=results_df["Accuracy"], y=results_df.index, palette="crest")
plt.title("Model Accuracy Comparison After ADASYN")
plt.xlabel("Accuracy Score")
plt.xlim(0, 1)
plt.tight_layout()
plt.show()

# Best model confusion matrix and report
best_model_name = results_df.index[0]
best_model = models[best_model_name]
y_pred_best = best_model.predict(X_test)

conf_matrix = confusion_matrix(y_test, y_pred_best)
plt.figure(figsize=(6, 5))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues')
plt.title(f"Confusion Matrix - {best_model_name}")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()

print(f"\nClassification Report for {best_model_name}:\n")
print(classification_report(y_test, y_pred_best))

print("Model Accuracy Scores After ADASYN:\n")
for model_name, acc in results.items():
    print(f"{model_name}: {acc:.4f}")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    AdaBoostClassifier
)
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from imblearn.over_sampling import SMOTE

# Assume df2 is already loaded and cleaned
# Define input features and target
X = df2.drop(columns=["accretion_rate_class_encoded", "river_name_encoded", "year"])
y = df2["accretion_rate_class_encoded"]

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

# Apply SMOTE to training data
smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

# Define models
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "AdaBoost": AdaBoostClassifier(random_state=42),
    "SVM": SVC(),
    "Naive Bayes": GaussianNB(),
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "XGBoost": XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=42)
}

# Train and evaluate models
results = {}

for name, model in models.items():
    model.fit(X_train_res, y_train_res)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    results[name] = acc

# Print accuracy results
print("Model Accuracy Scores After SMOTE:\n")
for model_name, acc in sorted(results.items(), key=lambda x: x[1], reverse=True):
    print(f"{model_name}: {acc:.4f}")
```

## ADASYN

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    AdaBoostClassifier
)
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from imblearn.over_sampling import ADASYN
from xgboost import XGBClassifier

# Assume df2 is already loaded and cleaned
X = df2.drop(columns=["accretion_rate_class_encoded", "river_name_encoded", "year"])
y = df2["accretion_rate_class_encoded"]

# Feature scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-test split with stratify
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

# Apply ADASYN to training data
adasyn = ADASYN(random_state=42)
X_train_res, y_train_res = adasyn.fit_resample(X_train, y_train)

# Define classifiers including AdaBoost and XGBoost
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "AdaBoost": AdaBoostClassifier(random_state=42),
    "SVM": SVC(),
    "Naive Bayes": GaussianNB(),
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "XGBoost": XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=42)
}

results = {}
reports = {}

# Train and evaluate each model on ADASYN-resampled data
for name, model in models.items():
    model.fit(X_train_res, y_train_res)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    results[name] = acc
    reports[name] = classification_report(y_test, y_pred, output_dict=True)

# Convert accuracy results to DataFrame
results_df = pd.DataFrame.from_dict(results, orient="index", columns=["Accuracy"]).sort_values(by="Accuracy", ascending=False)

# Plot model comparison
plt.figure(figsize=(10, 6))
sns.barplot(x=results_df["Accuracy"], y=results_df.index, palette="crest")
plt.title("Model Accuracy Comparison After ADASYN (Accretion)")
plt.xlabel("Accuracy Score")
plt.xlim(0, 1)
plt.tight_layout()
plt.show()

# Best model confusion matrix and report
best_model_name = results_df.index[0]
best_model = models[best_model_name]
y_pred_best = best_model.predict(X_test)

conf_matrix = confusion_matrix(y_test, y_pred_best)
plt.figure(figsize=(6, 5))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues')
plt.title(f"Confusion Matrix - {best_model_name} (Accretion)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()

print(f"\nClassification Report for {best_model_name}:\n")
print(classification_report(y_test, y_pred_best))

print("Model Accuracy Scores After ADASYN:\n")
for model_name, acc in results.items():
    print(f"{model_name}: {acc:.4f}")
```
